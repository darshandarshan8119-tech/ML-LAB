```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error

# 1. Dataset
data = {
    'X': [1, 2, 3, 4, 5, 6, 7, 8, 9],
    'Y': [5, 7, 9, 11, 13, 15, 17, 19, 21]
}
df = pd.DataFrame(data)

X = df[['X']]
y = df['Y']

# 2. Split data: 5 Train, 2 Validation, 2 Test
X_train, y_train = X.iloc[:5], y.iloc[:5]
X_val, y_val     = X.iloc[5:7], y.iloc[5:7]
X_test, y_test   = X.iloc[7:], y.iloc[7:]

# 3. Scale features (Important for Ridge and Lasso)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)

# 4. Initialize Models
models = {
    'OLS (Linear Regression)': LinearRegression(),
    'Ridge (lambda=1)': Ridge(alpha=1.0),
    'Lasso (lambda=1)': Lasso(alpha=1.0)
}

results = []

for name, model in models.items():
    if name == 'OLS (Linear Regression)':
        model.fit(X_train, y_train)
        train_pred = model.predict(X_train)
        val_pred = model.predict(X_val)
        test_pred = model.predict(X_test)
    else:
        model.fit(X_train_scaled, y_train)
        train_pred = model.predict(X_train_scaled)
        val_pred = model.predict(X_val_scaled)
        test_pred = model.predict(X_test_scaled)
    
    # Calculate MSE
    results.append({
        'Model': name,
        'Train MSE': mean_squared_error(y_train, train_pred),
        'Validation MSE': mean_squared_error(y_val, val_pred),
        'Test MSE': mean_squared_error(y_test, test_pred)
    })

# Display results
results_df = pd.DataFrame(results)
print(results_df.to_string(index=False))
```

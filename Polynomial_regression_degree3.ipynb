```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# Dataset
data = {
    'X': [-2, -1, 0, 1, 2],
    'Y': [-11, -2, 1, 4, 13]
}
df = pd.DataFrame(data)

X = df[['X']]
y = df['Y']

# Transform features for Polynomial Regression (degree = 3)
poly = PolynomialFeatures(degree=3)
X_poly = poly.fit_transform(X)

# Train the Linear Regression model
model = LinearRegression()
model.fit(X_poly, y)

# Extract and round coefficients to integers to avoid floating-point/negative-zero artifacts
d = int(round(model.intercept_))      # Intercept (constant term)
c = int(round(model.coef_[1]))        # Coefficient for X
b = int(round(model.coef_[2]))        # Coefficient for X^2
a = int(round(model.coef_[3]))        # Coefficient for X^3

print(f"a = {a}")
print(f"b = {b}")
print(f"c = {c}")
print(f"d = {d}")
print(f"\nPolynomial Equation: Y = {a}X^3 + {b}X^2 + {c}X + {d}")
```
